```python
trips = 25
avg_fare = 220.0
fuel_cost = 1100.0
rider_name ="kevin"

gross_income = trips*avg_fare
net_income = gross_income-fuel_cost
margin_pct = net_income/ gross_income *100

print(f'{rider_name}: gross_KES{gross_income:,.2f},'
f'net KES{net_income:,.2f},margin{margin_pct:.1f}%')
```

```python
inventory = [
    {"name": "Unga", "stock": 12, "reorder_level": 15},
    {"name": "Sugar", "stock": 40, "reorder_level": 20},
    {"name": "Milk", "stock": 8, "reorder_level": 10},
    {"name": "Bread", "stock": 25, "reorder_level": 10},
]

def restock_report(items):
    """Print stock status and return a list of products to reorder."""
    to_reorder = []
    for item in items:
        if item["stock"] <= item["reorder_level"]:
            status = "REORDER"
            to_reorder.append(item["name"])
        else:
            status = "OK"
        print(f"{item['name']:<6} stock={item['stock']:>3}  {status}")
    return to_reorder

print(restock_report(inventory))
```
